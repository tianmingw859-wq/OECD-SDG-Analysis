# OECD SDG Analysis – Reproducible Workflow

Input: OECD_SDG_Analysis.xlsx (Sheet1)

Scope: 38 OECD countries, 2000–2023

Indicators:
- sdg3_lifee
- sdg4_tertiary
- sdg8_unemp
- sdg10_gini

Required libraries:
pandas, numpy, scipy, matplotlib, statsmodels, openpyxl

Execution:
Run all notebook cells sequentially from a restarted kernel.

Note:
The Excel file is a prepared OECD extract.
The notebook reproduces the statistical analysis from this
input, not the extraction from the full source database.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_excel("OECD_SDG_Analysis.xlsx", sheet_name="Sheet1")

df.head()
```

```python

# Data quality and reproducibility checks

variables = [
    'sdg3_lifee',
    'sdg4_tertiary',
    'sdg8_unemp',
    'sdg10_gini'
]

required_columns = [
    'id', 'country', 'year', 'indexreg'
] + variables

# 1. Check required columns
missing_columns = [
    col for col in required_columns if col not in df.columns
]
assert not missing_columns, f"Missing columns: {missing_columns}"

# 2. Check dataset dimensions
print("Total observations:", len(df))
print("Number of countries:", df['country'].nunique())
print("Number of years:", df['year'].nunique())

# 3. Check OECD classification and year range
assert df['indexreg'].eq('OECD').all()
assert set(df['year']) == set(range(2000, 2024))

# 4. Check duplicate country-year records
duplicates = df.duplicated(
    subset=['id', 'year']
).sum()
print("Duplicate country-year records:", duplicates)
assert duplicates == 0

# 5. Check numeric data types
for col in ['year'] + variables:
    assert pd.api.types.is_numeric_dtype(df[col]), (
        f"{col} is not numeric"
    )

# 6. Check missing values
print("\nValid observations:")
print(df[variables].count())

print("\nMissing observations:")
print(df[variables].isna().sum())

# 7. Check complete observations across all four indicators
complete_rows = df[variables].notna().all(axis=1)
print("\nComplete four-indicator observations:",
      complete_rows.sum())

# 8. Verify the year with the highest complete coverage
annual_complete = (
    df.assign(complete=complete_rows)
      .groupby('year')['complete']
      .sum()
)

print("\nYear with highest complete coverage:",
      annual_complete.idxmax())

print("Complete observations in 2016:",
      annual_complete.loc[2016])
```

```python
trend_vars = {
    'sdg3_lifee': 'lifee',
    'sdg4_tertiary': 'tertiary',
    'sdg8_unemp': 'unemp',
    'sdg10_gini': 'gini'
}

results = []

for country, group in df.groupby('country'):
    row = {'country': country}

    for var, short_name in trend_vars.items():
        temp = group[['year', var]].dropna()

        if len(temp) >= 5:
            slope = np.polyfit(temp['year'], temp[var], 1)[0]
            row[f'{short_name}_slope'] = slope
            row[f'{short_name}_n'] = len(temp)
        else:
            row[f'{short_name}_slope'] = np.nan
            row[f'{short_name}_n'] = len(temp)

    results.append(row)

country_slopes = pd.DataFrame(results)

country_slopes.head()
```

```python
reliable_candidates = country_compare[
    (country_compare['lifee_n'] >= 15) &
    (country_compare['tertiary_n'] >= 15) &
    (country_compare['unemp_n'] >= 15) &
    (country_compare['gini_n'] >= 15)
].copy()

reliable_candidates[[
    'country',
    'lifee_n',
    'tertiary_n',
    'unemp_n',
    'gini_n'
]].sort_values('country')
```

```python
diff_cols = [
    'lifee_diff',
    'tertiary_diff',
    'unemp_diff',
    'gini_diff'
]

reliable_compare = reliable_candidates.copy()

for col in diff_cols:
    reliable_compare[col + '_z'] = (
        reliable_compare[col] / reliable_compare[col].std()
    )
reliable_compare['mean_abs_z'] = reliable_compare[
    [col + '_z' for col in diff_cols]
].abs().mean(axis=1)

reliable_compare[[
    'country',
    'lifee_diff_z',
    'tertiary_diff_z',
    'unemp_diff_z',
    'gini_diff_z',
    'mean_abs_z'
]].sort_values('mean_abs_z', ascending=False).head(10).round(3)
```

```python
all_patterns = reliable_compare[['country']].copy()

all_patterns['Life expectancy'] = np.where(
    reliable_compare['lifee_diff'] > 0, 'Better', 'Worse'
)

all_patterns['Tertiary education'] = np.where(
    reliable_compare['tertiary_diff'] > 0, 'Better', 'Worse'
)

all_patterns['Unemployment'] = np.where(
    reliable_compare['unemp_diff'] < 0, 'Better', 'Worse'
)

all_patterns['Gini inequality'] = np.where(
    reliable_compare['gini_diff'] < 0, 'Better', 'Worse'
)

indicator_cols = [
    'Life expectancy',
    'Tertiary education',
    'Unemployment',
    'Gini inequality'
]

all_patterns['Better_count'] = (
    all_patterns[indicator_cols] == 'Better'
).sum(axis=1)

all_patterns['Worse_count'] = (
    all_patterns[indicator_cols] == 'Worse'
).sum(axis=1)

all_patterns.sort_values(
    ['Better_count', 'country'],
    ascending=[False, True]
)
```

```python
notable_patterns = reliable_compare[['country']].copy()

# 把方向统一成：正值 = better，负值 = worse
notable_patterns['lifee_z'] = reliable_compare['lifee_diff_z']
notable_patterns['tertiary_z'] = reliable_compare['tertiary_diff_z']
notable_patterns['unemp_z'] = -reliable_compare['unemp_diff_z']
notable_patterns['gini_z'] = -reliable_compare['gini_diff_z']

def classify(z):
    if z >= 1:
        return 'Notably better'
    elif z <= -1:
        return 'Notably worse'
    else:
        return 'Similar to OECD'

notable_patterns['Life expectancy'] = notable_patterns['lifee_z'].apply(classify)
notable_patterns['Tertiary education'] = notable_patterns['tertiary_z'].apply(classify)
notable_patterns['Unemployment'] = notable_patterns['unemp_z'].apply(classify)
notable_patterns['Gini inequality'] = notable_patterns['gini_z'].apply(classify)

pattern_cols = [
    'Life expectancy',
    'Tertiary education',
    'Unemployment',
    'Gini inequality'
]

notable_patterns['Notable_count'] = (
    notable_patterns[pattern_cols] != 'Similar to OECD'
).sum(axis=1)

notable_patterns[[
    'country',
    'Life expectancy',
    'Tertiary education',
    'Unemployment',
    'Gini inequality',
    'Notable_count'
]].sort_values(
    ['Notable_count', 'country'],
    ascending=[False, True]
)
```

```python

# RQ3 sensitivity analysis 1:
# Compare three classification thresholds.
final_countries = [
    'Slovak Republic',
    'Costa Rica',
    'Estonia',
    'Luxembourg'
]

final_profiles = notable_patterns[
    notable_patterns['country'].isin(final_countries)
][[
    'country',
    'Life expectancy',
    'Tertiary education',
    'Unemployment',
    'Gini inequality',
    'Notable_count'
]].copy()

thresholds = [0.75, 1.00, 1.25]

indicator_z_cols = {
    'Life expectancy': 'lifee_z',
    'Tertiary education': 'tertiary_z',
    'Unemployment': 'unemp_z',
    'Gini inequality': 'gini_z'
}

def classify_threshold(z, threshold):
    if z >= threshold:
        return 'NB'
    elif z <= -threshold:
        return 'NW'
    else:
        return 'S'

sensitivity_rows = []

for country in final_countries:
    record = notable_patterns.loc[
        notable_patterns['country'] == country
    ].iloc[0]

    for indicator, z_col in indicator_z_cols.items():
        z = float(record[z_col])

        sensitivity_rows.append({
            'Country': country,
            'Indicator': indicator,
            'Standardised difference': round(z, 3),
            '±0.75': classify_threshold(z, 0.75),
            '±1.00': classify_threshold(z, 1.00),
            '±1.25': classify_threshold(z, 1.25)
        })

rq3_threshold_sensitivity = pd.DataFrame(sensitivity_rows)

display(rq3_threshold_sensitivity)

changes_075 = (
    rq3_threshold_sensitivity['±0.75']
    != rq3_threshold_sensitivity['±1.00']
).sum()

changes_125 = (
    rq3_threshold_sensitivity['±1.25']
    != rq3_threshold_sensitivity['±1.00']
).sum()

print("Number of classifications:", len(rq3_threshold_sensitivity))
print("Changes at ±0.75:", changes_075)
print("Changes at ±1.25:", changes_125)

assert len(rq3_threshold_sensitivity) == 16
assert changes_075 == 4
assert changes_125 == 0
```

```python

# RQ3 sensitivity analysis 2:
# Increase minimum valid observations from 15 to 20.

count_cols = [
    'lifee_n',
    'tertiary_n',
    'unemp_n',
    'gini_n'
]

reliable_20 = country_compare.loc[
    (country_compare[count_cols] >= 20).all(axis=1)
].copy()

# Recalculate standard deviations using eligible countries.
direction = {
    'lifee': 1,
    'tertiary': 1,
    'unemp': -1,
    'gini': -1
}

for indicator, sign in direction.items():
    diff_col = indicator + '_diff'
    z_col = indicator + '_z_20'

    reliable_20[z_col] = (
        sign * reliable_20[diff_col]
        / reliable_20[diff_col].std()
    )

coverage_rows = []

for country in final_countries:
    record = reliable_20.loc[
        reliable_20['country'] == country
    ].iloc[0]

    original = notable_patterns.loc[
        notable_patterns['country'] == country
    ].iloc[0]

    for indicator, short_name in zip(
        indicator_z_cols.keys(),
        direction.keys()
    ):
        original_col = indicator_z_cols[indicator]
        new_col = short_name + '_z_20'

        coverage_rows.append({
            'Country': country,
            'Indicator': indicator,
            '15-year rule': classify_threshold(
                original[original_col], 1.0
            ),
            '20-year rule': classify_threshold(
                record[new_col], 1.0
            )
        })

rq3_coverage_sensitivity = pd.DataFrame(coverage_rows)

print("Eligible countries at 15 years:", len(reliable_candidates))
print("Eligible countries at 20 years:", len(reliable_20))

display(rq3_coverage_sensitivity)

changes_20 = (
    rq3_coverage_sensitivity['15-year rule']
    != rq3_coverage_sensitivity['20-year rule']
).sum()

print("Classification changes:", changes_20)

assert len(reliable_candidates) == 30
assert len(reliable_20) == 26
assert changes_20 == 0
```

```python

# RQ2: Fixed-country sensitivity analysis

from scipy.stats import linregress

fixed_sample_results = []

for var in ['sdg4_tertiary', 'sdg10_gini']:
    counts = df.groupby('country')[var].count()
    eligible = counts[counts == 24].index

    fixed_median = (
        df[df['country'].isin(eligible)]
        .groupby('year')[var]
        .median()
    )

    original = linregress(
        annual_median.index,
        annual_median[var]
    )

    fixed = linregress(
        fixed_median.index,
        fixed_median.values
    )

    fixed_sample_results.append({
        'Indicator': var,
        'Complete_countries': len(eligible),
        'Original_slope': original.slope,
        'Fixed_slope': fixed.slope,
        'Original_R2': original.rvalue ** 2,
        'Fixed_R2': fixed.rvalue ** 2,
        'Fixed_p': fixed.pvalue
    })

rq2_fixed_sensitivity = pd.DataFrame(
    fixed_sample_results
)

display(rq2_fixed_sensitivity.round(4))
```

```python
from statsmodels.stats.multitest import multipletests

correlation_adjusted = correlation_results.copy()

correlation_adjusted['Pearson_FDR_p'] = multipletests(
    correlation_adjusted['Pearson p'],
    method='fdr_bh'
)[1]

correlation_adjusted['Spearman_FDR_p'] = multipletests(
    correlation_adjusted['Spearman p'],
    method='fdr_bh'
)[1]

correlation_adjusted[[
    'Variable 1',
    'Variable 2',
    'n',
    'Pearson r',
    'Pearson p',
    'Pearson_FDR_p',
    'Spearman rho',
    'Spearman p',
    'Spearman_FDR_p'
]].round(4)
```

```python
from scipy.stats import norm

ci_results = correlation_adjusted.copy()

lower_ci = []
upper_ci = []

for _, row in ci_results.iterrows():
    r = row['Pearson r']
    n = row['n']

    z = np.arctanh(r)
    se = 1 / np.sqrt(n - 3)
    z_crit = norm.ppf(0.975)

    lower = np.tanh(z - z_crit * se)
    upper = np.tanh(z + z_crit * se)

    lower_ci.append(lower)
    upper_ci.append(upper)

ci_results['Pearson_CI_lower'] = lower_ci
ci_results['Pearson_CI_upper'] = upper_ci

ci_results[[
    'Variable 1',
    'Variable 2',
    'n',
    'Pearson r',
    'Pearson_CI_lower',
    'Pearson_CI_upper',
    'Pearson p',
    'Pearson_FDR_p'
]].round(4)
```

```python
from itertools import combinations
from scipy.stats import pearsonr

yearly_corr_results = []

for year in sorted(df['year'].unique()):
    year_data = df[df['year'] == year]

    for var1, var2 in combinations(variables, 2):
        pair_data = year_data[[var1, var2]].dropna()

        if len(pair_data) >= 20:
            r, p = pearsonr(pair_data[var1], pair_data[var2])

            yearly_corr_results.append({
                'year': year,
                'Variable 1': var1,
                'Variable 2': var2,
                'n': len(pair_data),
                'Pearson r': r,
                'p_value': p
            })

yearly_correlations = pd.DataFrame(yearly_corr_results)

yearly_correlations.round(3)
```
